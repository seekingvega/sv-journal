## TradingView Screener Demo
while [PineScreener](https://www.tradingview.com/pine-screener/) is a premium feature in TradingView, the vanilla [Stock Screener](https://www.tradingview.com/etf-screener/pGOww70e/) is free
and now there's even a python package for it: [tradingview-screener](https://github.com/shner-elmo/TradingView-Screener)

So maybe this could be your first step in creating your own PineScreener with python!

All the filters available on TradingView are accessible. The package come with a very convenient search tool for it [here](https://shner-elmo.github.io/TradingView-Screener/fields/stocks.html).

As demo, below is my screener for:

* US ETFs
* non-OTC
* non-leveraged
* has NAV total return of over 5% ("risk-free" rate) for the last year
* has over $1mm dollar traded per day

```python
#| code-fold: false

from tradingview_screener import Query, col

screen = (Query()
    .select('name', 'exchange', 'close', 'volume', 'Perf.Y', 'nav_total_return.1Y', 'description')
    .where(
        col('typespecs').has(['etf']),
        col('leveraged_flag') == "Non-leveraged",
        col('exchange')!= 'OTC', 
        col("Value.Traded|1")>1000000,
        col('nav_total_return.1Y')> 0.05
        )
    .order_by('Perf.Y', ascending=False)
    .limit(10))
screen.set_markets('america')

r = screen.get_scanner_data()
df_screen = r[1]
df_screen
```
